```python
import numpy as np
from abc import ABC,abstractmethod
```

```python
class Layers(ABC):
    @abstractmethod
    def forward_propogation(self,inputs):
        pass
```

```python
class Neural_Network(Layers):
    def __init__(self,inputs,neurons):
        self.weights=np.random.randn(inputs,neurons)
        self.bias=np.ones(neurons)
    def forward_propogation(self,inputs):
        self.outputs=np.dot(inputs,self.weights)+self.bias
```

```python
class ReLu(Layers):
    def forward_propogation(self,inputs):
        self.output=np.maximum(0,inputs)
```

```python
class sigmoid(Layers):
    def forward_propogation(self,inputs):
        self.output=1/(1+np.exp(-inputs))
```

```python
class categorical_cross_entropy_loss:
    def forward(self, y_pred, y_true):

        # Prevent log(0)
        y_pred = np.clip(y_pred, 1e-7, 1 - 1e-7)

        samples = len(y_pred)

        # If labels are one-hot encoded
        if len(y_true.shape) == 2:
            y_true = np.argmax(y_true, axis=1)

        correct_confidence = y_pred[
            range(samples),
            y_true
        ]

        negative_log_likelihood = -np.log(correct_confidence)

        return np.mean(negative_log_likelihood)


# ---------------- SAMPLE DATA ---------------- #

X = np.array([
    [1, 2, 3, 2],
    [2, 3, 4, 1],
    [3, 4, 5, 2]
])

y_true = np.array([0, 1, 1])


# ---------------- NETWORK FLOW ---------------- #

# Input Layer → Hidden Layer
dense1 = Neural_Network(4, 5)
dense1.forward_propogation(X)

print("Dense Layer Output:\n")
print(dense1.outputs)


# ReLU Activation
activation1 = ReLu()
activation1.forward_propogation(dense1.outputs)

print("\nReLU Output:\n")
print(activation1.output)


# Output Layer
dense2 = Neural_Network(5, 2)
dense2.forward_propogation(activation1.output)

print("\nSecond Dense Layer Output:\n")
print(dense2.outputs)


# Sigmoid Activation
activation2 = sigmoid()
activation2.forward_propogation(dense2.outputs)

print("\nSigmoid Output:\n")
print(activation2.output)


# Loss Calculation
loss_function = categorical_cross_entropy_loss()

loss = loss_function.forward(
    activation2.output,
    y_true
)

print("\nCategorical Cross Entropy Loss:\n")
print(loss)
```
